```python
%sql
create or replace function dev.vivek.ranking_rich(net_worth_in_billions integer)
returns string
return case when net_worth_in_billions > 100 then 'HIGH' when net_worth_in_billions > 50 then 'MED' else 'LOW' end
```

```python
%sql

select *,dev.vivek.ranking_rich(net_worth_in_billions) as category from dev.vivek.richest_people
```
